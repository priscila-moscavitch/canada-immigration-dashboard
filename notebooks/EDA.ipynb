### Creating formatted labels for visualization

Some country names are too long and affect the readability of visualizations.  
To improve clarity, a shortened version of country names will be created for display purposes.

```python
df['Country_short'] = df['Country'].replace({
    'United Kingdom of Great Britain and Northern Ireland': 'United Kingdom',
    'Iran (Islamic Republic of)': 'Iran'
})
```

## 8. Distribution of total immigration

This histogram shows how immigration values are distributed across countries.

It helps identify concentration and possible outliers.

```python
df_filtered = df[df['Total'] < 100000]

fig = px.histogram(
    df_filtered,
    x='Total',
    nbins=40,
    title='Distribution of Total Immigration to Canada',
    labels={
        'Total': 'Number of Immigrants',
        'count': 'Number of Countries'
    }
)

fig.update_layout(
    title_x=0.5,
    xaxis_title='Number of Immigrants',
    yaxis_title='Number of Countries',
    template='plotly_white'
)

fig.show()
```

## Key insights 

This histogram shows how immigration values are distributed across countries.

To improve visualization, extreme values were filtered out, as they can distort the distribution.

Most countries contribute relatively small numbers of immigrants, while a few countries have significantly higher values.

This creates a right-skewed distribution, with a clear concentration in lower ranges and a long tail of higher values, indicating the presence of outliers.

## 9. Comparing immigration over time

This scatter plot compares immigration numbers between 1980 and 2013.

It helps identify growth patterns and country trends over time.

```python
fig = px.scatter(
    df,
    x='1980',
    y='2013',
    hover_name='Country',
    color='Continent',
    title='Immigration to Canada: 1980 vs 2013'
)

fig.show()
```

## Key insights

This scatter plot compares immigration numbers between 1980 and 2013.

Most countries show higher immigration levels in 2013 than in 1980, indicating an overall growth trend over time.

Asian countries stand out with the highest increases, while other regions show more moderate changes.

The distribution also suggests a positive correlation between past and recent immigration levels, meaning countries with higher immigration in 1980 tend to maintain higher levels in 2013.

## 10. Top 10 Countries by Total Immigration

Analysis of the countries with the highest total number of immigrants.

This helps highlight the main contributors to immigration to Canada.

```python
top_10 = df.sort_values(by='Total', ascending=False).head(10)

fig = px.bar(
    top_10,
    x='Country_short',
    y='Total',
    color='Continent',
    title='Top 10 Countries by Total Immigration to Canada'
)

fig.show()
```

## Key Insights

- Immigration is highly concentrated among a few countries
- There is a clear upward trend over time
- Regional differences are significant, especially in Asia

## 11. Conclusion

This analysis reveals that immigration to Canada is not evenly distributed across countries.

A small number of countries contribute disproportionately to total immigration, while most countries have relatively low numbers.

Over time, there is a clear upward trend, suggesting increasing global mobility and stronger migration flows.

These patterns highlight the importance of regional dynamics and policy impacts on immigration trends.

Further analysis could explore the socio-economic or political factors driving these patterns.
